### Importar bases de dados (Excel)
### Criar os DataFrames referentes à cada base

```python
import pandas as pd

# importar arquivos e criar DataFrames

df_funcionarios = pd.read_excel("funcionarios.xlsx")
df_historico_salarios = pd.read_excel("historico_salarios.xlsx")
df_setores = pd.read_excel("setores.xlsx")
```

```python
df_funcionarios.head()
```

### Exibir Nome e Salário da base de funcionários

```python
print(df_funcionarios[["Nome", "Salario"]])
```

### Filtrar funcionários com salário acima de 5 mil

```python
print(df_funcionarios[df_funcionarios["Salario"] > 5000])

```

### Merge entre funcionários e setores (juntar tabelas)

```python
# União das duas tabelas em uma apenas - tem que usar uma coluna de referência
# vamos usar o ID, como referência, porque tem nas duas tabelas

df_funcionarios_setor = df_funcionarios.merge(df_setores, on="IdSetor")
```

```python
# para ver a tabela nova

df_funcionarios_setor
```

```python
# exibir as 3 colunas nome, setor e salário
df_funcionarios_setor[["Nome", "Setor", "Salario"]]
```

### Adicionar novos funcionários (concat)

```python
# fazer um dicionário criando uma nova variável (novos funcionários)

novos_funcionarios = {
    "IdFuncionario":[7, 8],
    "Nome": ["Gabriel", "Helena"],
    "IdSetor": [3, 4],
    "Salario": [4500, 5300],
    "Idade": [31, 38],
    }

df_novos_funcionarios = pd.DataFrame(novos_funcionarios)

```

```python
df_novos_funcionarios
```

```python
df_funcionarios = pd.concat([df_funcionarios, df_novos_funcionarios], ignore_index= True)
```

```python
df_funcionarios
```

```python
df_resultado = df_funcionarios.merge(df_setores, on="IdSetor")
```

```python
print(df_resultado[["Nome", "Setor", "Salario"]])
```

```python
print(df_funcionarios)
```

### Agrupando valores

```python
# usar a funçao mean para saber a média salarial de cada setor, o groupby foi p apresentar os valores

print(df_funcionarios_setor.groupby("Setor")["Salario"].mean())
```
